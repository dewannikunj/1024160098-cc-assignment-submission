```python
n = int(input("Enter a number: "))

sum = 0

for i in range(2, n + 1):
    count = 0

    for j in range(1, i + 1):
        if i % j == 0:
            count = count + 1

    if count == 2:
        sum = sum + i

print("Sum of prime numbers =", sum)
```

```python
def add_odd(n):
    sum = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            sum = sum + i

    return sum


n = int(input("Enter a number: "))

result = add_odd(n)

print("Sum of odd numbers =", result)
```

```python
def add_prime(n):
    sum = 0

    for i in range(2, n + 1):
        count = 0

        for j in range(1, i + 1):
            if i % j == 0:
                count = count + 1

        if count == 2:
            sum = sum + i

    return sum


n = int(input("Enter a number: "))

result = add_prime(n)

print("Sum of prime numbers =", result)
```
