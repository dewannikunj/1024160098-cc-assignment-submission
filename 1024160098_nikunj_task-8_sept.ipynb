Q4. Hypothesis generation should be written before looking at test results. For this student-performance problem, generate and test the following:
a. State one research problem for this database.
b. Write H0 and H1 for: students who attended extra classes have a higher mean exam score than students who did not.
c. This is a small sample (n < 30). Compute a two-sample t-statistic using numpy only:
   t = (mean_yes - mean_no) / sqrt(s_yes^2 / n_yes + s_no^2 / n_no)
d. Use significance level alpha = 0.05. For a one-tailed test with small samples, compare |t| with ttab = 1.812 (df ≈ 10). Make a decision: reject or do not reject H0.
e. Explain Type I and Type II error in the context of this extra-class hypothesis.


```python
print("a. Research problem: Do extra classes improve exam scores of students?")

print("b. H0: Extra-class students do not have a higher mean exam score than other students.")
print("   H0: mu_yes <= mu_no")
print("   H1: Extra-class students have a higher mean exam score than other students.")
print("   H1: mu_yes > mu_no")

yes_scores = df.loc[df["extra_class"] == "Yes", "exam_score"].to_numpy(dtype=float)
no_scores = df.loc[df["extra_class"] == "No", "exam_score"].to_numpy(dtype=float)

mean_yes = np.mean(yes_scores)
mean_no = np.mean(no_scores)
s_yes = np.std(yes_scores, ddof=1)
s_no = np.std(no_scores, ddof=1)
n_yes = yes_scores.size
n_no = no_scores.size

t_cal = (mean_yes - mean_no) / np.sqrt((s_yes ** 2) / n_yes + (s_no ** 2) / n_no)
alpha = 0.05
t_tab = 1.812

print("c. Extra class = Yes scores:", yes_scores)
print("   Extra class = No scores:", no_scores)
print("   mean_yes =", mean_yes, "mean_no =", mean_no)
print("   s_yes =", s_yes, "s_no =", s_no)
print("   n_yes =", n_yes, "n_no =", n_no)
print("   t_cal =", t_cal)

print("d. alpha =", alpha)
print("   t_tab (one-tailed, df ~ 10) =", t_tab)
print("   t_cal > t_tab:", t_cal > t_tab)
if t_cal > t_tab:
    print("   Decision: Reject H0. The sample supports that extra classes are associated with a higher mean exam score.")
else:
    print("   Decision: Do not reject H0. There is not enough evidence that extra classes increase mean exam score.")

print(
    "e. Type I error (alpha): concluding that extra classes improve scores when they actually do not "
    "(rejecting a true H0). Type II error (beta): concluding that extra classes do not help when they actually do "
    "(failing to reject a false H0)."
)
```

Q5. A simple cognitive-assistive system wants to predict exam_score from three features: sleep_hours, study_hours and stress_level.
X contains these three columns from the database. y is exam_score.
a. Display the shape and dimensions of X.
b. Calculate X.T and explain what the transpose represents.
c. Calculate the matrix product X.T @ X.
d. Calculate the matrix inverse of X.T @ X using np.linalg.inv().
e. Implement the Ordinary Least Squares equation: beta = (X.T @ X)^-1 @ X.T @ y
f. Explain what the resulting coefficients represent for the three features.
g. A new student has: new_student = np.array([6, 7, 4]). Use the coefficients to obtain a predicted exam score.


```python
X = df[["sleep_hours", "study_hours", "stress_level"]].to_numpy(dtype=float)
y = df["exam_score"].to_numpy(dtype=float)

print("X:")
print(X)
print("y:", y)

print("a. Shape of X:", X.shape)
print("   Dimensions (ndim) of X:", X.ndim)

print("b. X.T:")
print(X.T)
print("   Transpose swaps rows and columns. X is 12 students x 3 features. X.T is 3 features x 12 students.")

XtX = X.T @ X
print("c. X.T @ X:")
print(XtX)

XtX_inv = np.linalg.inv(XtX)
print("d. Inverse of X.T @ X:")
print(XtX_inv)

beta = XtX_inv @ X.T @ y
print("e. OLS coefficients beta:")
print(beta)

print(
    "f. The three coefficients correspond to sleep_hours, study_hours and stress_level. "
    "Each value shows how much the predicted exam score changes when that feature increases by 1 unit, "
    "holding the other features constant."
)
print("   sleep_hours coefficient:", beta[0])
print("   study_hours coefficient:", beta[1])
print("   stress_level coefficient:", beta[2])

new_student = np.array([6, 7, 4], dtype=float)
predicted = new_student @ beta
print("g. new_student:", new_student)
print("   Predicted exam score:", predicted)
```
