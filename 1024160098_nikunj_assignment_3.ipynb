6. Create a sample dataset (employees.csv) containing information about employees in a company.

WAP to determine the following using dataset in Q6:

a) Shape (number of rows and columns) of the DataFrame.

b) Summary of the DataFrame that includes the data types and non-null counts for each column.

c) Generate descriptive statistics.

d) Display the first 5 rows and last 3 rows of the dataset.

e) Calculate the following statistics from the dataset:
i. The average salary of employees.
ii. The total bonus paid to all employees.
iii. The youngest employee's age.
iv. The highest performance rating.

f) Sort the DataFrame by the Salary column in descending order.

g) Add a new column that categorizes employees based on their performance rating:
i. Excellent for ratings >= 4.5
ii. Good for ratings >= 4.0 but < 4.5
iii. Average for ratings < 4.0

h) Identify missing values in the DataFrame.

i) Rename the Employee_ID column to ID.

j) Find all employees who:
i. Have more than 5 years of experience.
ii. Belong to the IT department.

k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.

l) Save the modified DataFrame (with added columns) to a new CSV file.


```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Nikunj", "Mokksh", "Aryan", "Ananya", "Kabir"],
    "Department": ["IT", "HR", "IT", "Marketing", "Sales"],
    "Age": [29, 33, 40, 27, 37],
    "Salary": [52000, 71000, 68000, 54000, 61000],
    "Years_of_Experience": [4, 7, 11, 3, 12],
    "Joining_Date": [
        "2020-04-12",
        "2018-01-22",
        "2014-08-05",
        "2021-06-18",
        "2011-09-30"
    ],
    "Gender": ["Male", "Male", "Male", "Female", "Male"],
    "Bonus": [5200, 6800, 6100, 4600, 5100],
    "Rating": [4.5, 4.2, 3.7, 4.8, 3.6]
}

employees = pd.DataFrame(data)
employees.to_csv("employees.csv", index=False)
print(employees)
```

```python
print(employees.shape)
```

```python
employees.info()
```

```python
print(employees.describe())
```

```python
print(employees.head(5))
print()
print(employees.tail(3))
```

```python
average_salary = employees["Salary"].mean()
print("Average Salary =", average_salary)

total_bonus = employees["Bonus"].sum()
print("Total Bonus =", total_bonus)

youngest_age = employees["Age"].min()
print("Youngest Age =", youngest_age)

highest_rating = employees["Rating"].max()
print("Highest Performance Rating =", highest_rating)
```

```python
employees_sorted = employees.sort_values(by="Salary", ascending=False)
print(employees_sorted)
```

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Category"] = employees["Rating"].apply(categorize_rating)
print(employees[["Name", "Rating", "Category"]])
```

```python
print(employees.isnull().sum())
```

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)
print(employees)
```

```python
experienced_employees = employees[employees["Years_of_Experience"] > 5]
print("Employees with more than 5 years of experience")
print(experienced_employees)
print()
it_employees = employees[employees["Department"] == "IT"]
print("Employees in IT Department")
print(it_employees)
```

```python
employees["Tax"] = employees["Salary"] * 0.10
print(employees)
```

```python
employees.to_csv("employees_modified.csv", index=False)
print("Modified DataFrame saved as employees_modified.csv")
```
